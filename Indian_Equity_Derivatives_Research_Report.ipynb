## 5. Equity Factor Analysis

A systematic multi-factor framework was developed to rank a representative
universe of 20 large-cap Indian equities.

The model combines three dimensions:

1. **Quality — 35%**
2. **Valuation — 25%**
3. **Momentum — 40%**

Each factor is converted into a cross-sectional percentile score before being
combined into a composite equity score.

### 5.1 Quality Factor

The quality factor evaluates companies using:

- Return on Equity (ROE)
- Profit Margin

Each metric is ranked cross-sectionally across the equity universe.

The quality score is calculated as:

**55% ROE Score + 45% Profit Margin Score**

Missing ROE observations were not imputed as zero. Where ROE was unavailable,
the quality score was calculated using the available quality information.

### 5.2 Valuation Factor

Relative valuation was assessed using:

- Price-to-Earnings (P/E)
- Price-to-Book (P/B)

Lower valuation multiples receive higher percentile scores.

The valuation factor is calculated as:

**60% P/E Score + 40% P/B Score**

The factor is intended as a relative cross-sectional valuation measure rather
than a standalone indicator of investment attractiveness.

### 5.3 Momentum Factor

Price momentum was measured over three horizons:

- 3 months
- 6 months
- 12 months

The weighted momentum score assigns:

**30% to 3-month momentum**

**30% to 6-month momentum**

**40% to 12-month momentum**

Momentum was then adjusted for recent 60-day annualized volatility to obtain a
risk-adjusted momentum measure.

### 5.4 Composite Equity Score

The final equity score combines the three factors:

| Factor | Weight |
|---|---:|
| Quality | 35% |
| Valuation | 25% |
| Risk-adjusted Momentum | 40% |
| **Total** | **100%** |

The resulting composite score provides a transparent and systematic method for
ranking the selected equity universe.

## 7. Systematic Trading Strategy

A rules-based, long-only momentum strategy was developed to evaluate whether
simple trend-following signals could improve the risk characteristics of a
NIFTY 50 investment.

The strategy uses two moving averages together with a medium-term return filter.

### Entry Rule

A long position is established when:

- 20-day moving average > 50-day moving average
- 20-day NIFTY 50 return > 0

### Exit Rule

The position is exited when either condition is no longer satisfied.

When no position is held, the strategy remains in cash.

### Look-Ahead Bias Control

Trading signals are shifted forward by one trading day. Therefore, a signal
generated using information available at the close of day *t* is applied to
the return on day *t+1*.

This prevents the backtest from using information that would not have been
available when the simulated trade was executed.

### 7.1 Strategy Construction

The strategy can be represented conceptually as:

**Market Data → Moving Averages → Momentum Filter → Trading Signal → Position → Return**

The strategy is intentionally simple. The objective is not to optimize a large
number of parameters, but to evaluate whether a transparent and reproducible
trend-following rule can provide useful risk control.

The cumulative performance comparison illustrates the trade-off between

```python
from IPython.display import Image, display

display(Image("/content/strategy_vs_nifty.png"))
```

### 7.2 Market Exposure

The strategy was invested for approximately **49.5% of trading days** during
the research period.

This relatively low exposure is an important characteristic of the strategy.
The system moves to cash when its trend and momentum conditions are not met,
which contributes to its lower volatility and drawdown relative to the
buy-and-hold benchmark.

### 7.3 Strategy Behaviour Across Market Regimes

The strategy displayed substantial regime dependence.

During bearish periods, the strategy generally remained less exposed and
experienced smaller losses than the NIFTY 50 benchmark.

During bullish periods, the strategy participated in market gains but captured
considerably less upside than the benchmark.

This indicates that the strategy should be interpreted primarily as a
risk-control and trend-following framework rather than a strategy designed
to maximize absolute returns.

### 7.4 Look-Ahead Bias Control

To reduce look-ahead bias, each trading signal was shifted by one trading day.
Information used to generate a signal at the close of day t was therefore applied
only to the return observed on day t+1.

This ensures that the backtest does not assume execution using information that
would only become available after the trading decision.

### 7.5 Strategy Rationale

The strategy was intentionally designed as a simple trend-following framework.
The 20-day and 50-day moving averages capture medium-term trend direction, while
the 20-day return filter provides an additional momentum condition.

The objective was not to optimize a large number of parameters or maximize
historical returns. Instead, the strategy was designed to be transparent,
reproducible and suitable for evaluating the relationship between trend signals,
market regimes and portfolio risk.

The strategy moves to cash when its conditions are not satisfied, allowing us to
evaluate whether dynamic exposure can reduce drawdown and volatility relative
to a continuously invested NIFTY 50 benchmark.

## 9. Transaction Costs & Parameter Robustness

A systematic strategy can appear attractive in a historical backtest while
being sensitive to trading costs or arbitrary parameter choices. Two
additional tests were therefore performed:

1. A transaction-cost sensitivity test
2. A moving-average parameter robustness test

These tests assess whether the observed strategy characteristics remain
meaningful after accounting for implementation costs and alternative
strategy specifications.

### 9.1 Transaction Cost Analysis

A transaction cost of **10 basis points (0.10%)** was applied whenever the
strategy changed its position.

| Metric | Before Costs | After 10 bps Costs |
|---|---:|---:|
| CAGR | 5.56% | 3.56% |
| Annualized Volatility | 9.02% | 9.03% |
| Sharpe Ratio | 0.664 | 0.444 |
| Sortino Ratio | 0.634 | 0.435 |
| Maximum Drawdown | -16.21% | -17.04% |
| Win Rate | 55.29% | 54.91% |

Transaction costs materially reduced the strategy's risk-adjusted
performance, with CAGR falling from 5.56% to 3.56% and Sharpe ratio
declining from 0.664 to 0.444.

However, the strategy retained a relatively low volatility and drawdown
profile after costs.

The transaction-cost assumption is simplified and does not incorporate
broker-specific charges, taxes, market impact or bid-ask spread variation.

### 9.2 Moving-Average Parameter Robustness

To determine whether the strategy's behaviour depended on a single
moving-average specification, three parameter combinations were tested.

| Moving Average | CAGR | Annualized Volatility | Sharpe Ratio | Maximum Drawdown |
|---|---:|---:|---:|---:|
| 10 / 30 | 4.81% | 9.83% | 0.542 | -21.82% |
| 20 / 50 | 5.56% | 9.02% | 0.664 | -16.21% |
| 50 / 100 | 7.06% | 7.92% | 0.922 | -11.41% |

The strategy maintained a broadly defensive risk profile across all three
parameter combinations.

The 50 / 100 specification produced the highest CAGR and Sharpe ratio and
the lowest maximum drawdown among the tested configurations.

These results provide evidence that the strategy's general risk-reduction
characteristics are not unique to the baseline 20 / 50 configuration,
although the limited parameter grid does not constitute a comprehensive
robustness analysis.

### 9.3 Interpretation

The robustness analysis supports the interpretation that the strategy's
primary characteristic is reduced market exposure and downside risk rather
than persistent benchmark outperformance.

The results remain directionally consistent across different moving-average
parameters, while transaction costs reduce performance but do not eliminate
the strategy's defensive characteristics.

Further research would require walk-forward testing, a broader parameter
search and point-in-time fundamentals before drawing stronger conclusions
about out-of-sample performance.

## 11. Limitations

### 11.1 Fundamental Data Limitation

The equity fundamental variables used in the multi-factor model represent a
current snapshot rather than a point-in-time historical dataset.

Therefore, the quality and valuation scores were used for current cross-sectional
equity research and were **not** used to claim historical backtested portfolio
performance.

A point-in-time fundamental database would be required to rigorously test
whether the multi-factor model generated historical excess returns without
look-ahead bias.

### 11.2 Derivatives Data Limitation

The derivatives analysis is based on a NIFTY options-chain snapshot for the
08-Sep-2026 expiry rather than a complete historical options database.

As a result, the open-interest, implied-volatility and put-call ratio analysis
provides a snapshot of market positioning rather than a historical test of
derivatives-based trading signals.

A longer historical options dataset would be required to evaluate how
derivatives positioning behaves across different market regimes over time.

### 11.3 Strategy Scope

The backtested strategy is a long-only trend-following strategy and does not
model short selling, leverage or complex derivatives strategies.

The strategy is intentionally simple and uses a limited number of technical
signals. More sophisticated approaches may produce different results.

### 11.4 Transaction Cost & Execution Assumptions

Transaction costs were modelled using a simplified 10-basis-point assumption
applied when the strategy changed position.

The analysis does not fully model brokerage charges, taxes, bid-ask spreads,
slippage, market impact or execution delays.

Actual implementation costs may therefore differ from the assumptions used
in the backtest.

### 11.5 Parameter Testing

The robustness analysis evaluated three moving-average configurations:
10 / 30, 20 / 50 and 50 / 100.

Although the strategy's defensive characteristics persisted across these
specifications, the limited parameter grid does not constitute a comprehensive
optimization or walk-forward validation process.

Further testing across different market periods and parameter ranges would
provide stronger evidence of out-of-sample robustness.

### 11.6 Statistical Interpretation

The observed relationship between India VIX and future realized volatility
represents statistical association rather than causal proof.

The correlation and regression results should therefore be interpreted as
evidence of predictive association within the sample rather than definitive
evidence of a causal mechanism.

Similarly, historical backtest performance should not be interpreted as a
guarantee of future returns.

### 11.7 Data & Sample Limitations

The equity analysis uses a representative sample of 20 large-cap Indian
equities rather than the complete NIFTY 50 universe.

The derivatives analysis also focuses on a single options expiry.

These choices were made to maintain a practical and reproducible research
scope but limit the generalizability of the findings.
